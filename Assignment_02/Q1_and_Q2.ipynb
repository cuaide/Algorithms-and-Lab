```python
class Node:
  def __init__(self,name:str,grade:str,marks:int,gpa:float):
    self.name = name
    self.grade = grade
    self.marks = marks
    self.gpa = gpa
    self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def InsertNode(self, name: str, grade: str, marks: int, gpa: float):
        newNode = Node(name, grade, marks, gpa)
        if self.head is None:
            self.head = newNode
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = newNode
        return newNode

    def FindNode(self, name: str):
        current = self.head
        index = 0
        while current:
            if current.name == name:
                return index
            current = current.next
            index += 1
        return -1

    def DeleteNode(self, name: str):
        if self.head is None:
            return -1
        if self.head.name == name:
            self.head = self.head.next
            return 0
        current = self.head
        index = 0
        while current.next:
          index += 1
          if current.next.name == name:
            current.next = current.next.next
            return index
          current = current.next
        return -1
        print(f"{name} not found in list.")

    def DisplayList(self):
        current = self.head
        while current:
            print(f"Name: {current.name}, Grade: {current.grade}, Marks: {current.marks}, GPA: {current.gpa}")
            current = current.next
```

Create a Stack class that inherits from the List class. Modify the List class to allow the Stack class to access the private members of the List class.
Provide function prototypes and definitions for the following operations:
• Top: This function should receive no arguments and return the student's name of the first node (top node) in the stack.
• Push: This function should receive data to be stored as a node, create a node to store the data, and add the node as the first element of the linked list (on top of the stack).
• Pop: This function should receive no arguments, delete the first node (top node) of the stack, and return the student's name stored in the deleted node.

```python
class Stack(LinkedList):
    def top(self):
        if self.head is None:
            return None
        return self.head.name

    def push(self, name: str, grade: str, marks: int, gpa: float):
        newNode = Node(name, grade, marks, gpa)
        newNode.next = self.head
        self.head = newNode

    def pop(self):
        if self.head is None:
            return None
        pop_name = self.head.name
        self.head = self.head.next
        return pop_name
```

```python
s = Stack()
s.push('Alice','A',95,4.0)
s.push('Bob','B',85,3.5)
s.push('Carol','C',75,3.0)
s.push('David','A',92,3.8)

s.DisplayList()
```
